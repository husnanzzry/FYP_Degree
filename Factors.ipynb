```python
# Importing essential libraries and functions

import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords
from numpy import array
import tensorflow as tf
from numpy.random import seed
seed(2899)
tf.random.set_seed(99)
from sklearn.model_selection import train_test_split
```

```python
# Importing dataset

dataset = pd.read_csv("extractsentiment_vader.csv")
```

```python
dataset.head()
```

```python
# Define the keywords for each category
server_performance_keywords = ["load","lag","smooth", "down”, “cant open”, “cannot open”, “loading”, “slow","server",
                              "long”, time", "down", "stuck", "speed”, “error”, “wait”, “traffic”, “mainten”, “crash”,
                               "glitch”, “internet”, “speed”, “wifi”, “patient”, “patience”, “problem”, “hour”, “prob”, “steady”,
                               "lambat”, “laju”, “lancar”, “sabar”, “lembab”, “tunggu berjam”, “tunggu lama”, “kesabaran”, “cepat”,
                               "test sabar” “kerap down”, “apps rosak”, “transfer lambat”, “loading lama”, “loading laju”]
security_data_privacy_keywords = ["safe", "dangerous","deposit", "data privacy", "privacy", "secure", "security", "password",
                                  "scam","hack","encrypt","lost","data","passcode", "block", "policy","breach", "assistance","internal",
                                  "reliable", "checkout","expire","transfer", "transact","tac", "track","gone", "victim",
                                  "authent", "alert","otp","malware", "attack", "system","access to device","limit","expense",
                                  "advise","contactless","spend", "contact", "verify", "payment”,”secret”,“bahaya”, “rasa selamat”,
                                  "selamat”, “tak selamat”, “kena block”, “kena hack”, “history hilang”, “balance hilang”, “balance kurang”]
ui_ux_keywords = ["friendly", "button", "navigate","ui","confus","interface","design","responsive", "feature",
                  "hassle", "inefficient","restart", "difficult","apply”, bigpay", "store", "cute", "prize", "withdraw",
                  "facilitate","convenien",  "claim","cashless", "simple","setup","fun","enjoy","servic","qrpay","duitnow",
                  "launch","popular","upgrade","improve", "interactive","register","cashback","fix","easy","favourite",
                  "notif","futuristic","tng","shopee","ewallet", "senang guna”, “mesra pengguna”, “design menarik”,”design lawa”,
                   "design buruk”, “interface buruk”, “interface lawa”, “interface menarik”, “login senang”, “sign up senang”,
                    "sign up susah”, “login susah”, “interface serabut”, “interface semak”, “design semak”, “design serabut”, “feature bersepah”]

def categorize_text(text):
    text = text.lower()  # Convert text to lower case for case-insensitive matching
    if any(word in text for word in security_data_privacy_keywords):
        return "Security and Data Privacy"
    elif any(word in text for word in ui_ux_keywords):
        return "UI/UX"
    elif any(word in text for word in server_performance_keywords):
        return "Server Performance"
    else:
        return "Server Performance"

# Apply the categorize_text function to the 'text_lemmatized' column
dataset['Factors'] = dataset['text_lemmatized'].apply(categorize_text)

# Display the first few rows to verify
dataset.head(100)
```
